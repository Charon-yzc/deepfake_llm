```python
import json

def modify_json_paths(input_file, output_file):
    with open(input_file, 'r') as f:
        data = json.load(f)
    
    # 假设data是一个列表
    for item in data:
        if 'imagepath' in item:
            item['imagepath'] = item['imagepath'].replace('/mnt/d/real_syntheies_datasets/rgb/', '/Users/charon/Documents/')
        
    with open(output_file, 'w') as f:
        json.dump(data, f, indent=4)

# 使用示例
modify_json_paths('output/Deepfakes_analysis_results_mini.json', 'Deepfakes_analysis_results.json')
```
